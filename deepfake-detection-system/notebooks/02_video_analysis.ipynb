```python
# Create data exploration notebook using nbformat
%pip install nbformat

import nbformat
from nbformat.v4 import new_notebook, new_code_cell, new_markdown_cell

nb = new_notebook(
    cells=[
        new_markdown_cell("# Video Data Analysis\nExplore and analyze our video datasets"),
        new_code_cell(
            "# Setup\n"
            "import sys\n"
            "sys.path.append('../src')\n"
            "\n"
            "import cv2\n"
            "import numpy as np\n"
            "import pandas as pd\n"
            "import matplotlib.pyplot as plt\n"
            "from pathlib import Path\n"
            "from preprocessing.video_utils import VideoProcessor\n"
            "\n"
            "# Configure matplotlib\n"
            "%matplotlib inline\n"
            "plt.style.use('default')\n"
            "\n"
            "processor = VideoProcessor()\n"
            "print('✅ Video processor ready!')"
        ),
        new_code_cell(
            "# Analyze sample dataset\n"
            "real_dir = Path('../data/raw/sample_dataset/real')\n"
            "fake_dir = Path('../data/raw/sample_dataset/fake')\n"
            "\n"
            "real_videos = processor.find_videos(real_dir)\n"
            "fake_videos = processor.find_videos(fake_dir)\n"
            "\n"
            "print(f'Real videos found: {len(real_videos)}')\n"
            "print(f'Fake videos found: {len(fake_videos)}')\n"
            "\n"
            "# Analyze all videos\n"
            "video_data = []\n"
            "\n"
            "for video_path in real_videos:\n"
            "    info = processor.read_video_info(video_path)\n"
            "    info['label'] = 'real'\n"
            "    video_data.append(info)\n"
            "\n"
            "for video_path in fake_videos:\n"
            "    info = processor.read_video_info(video_path)\n"
            "    info['label'] = 'fake'\n"
            "    video_data.append(info)\n"
            "\n"
            "# Create DataFrame\n"
            "df = pd.DataFrame(video_data)\n"
            "print('\\nDataset Overview:')\n"
            "print(df[['filename', 'label', 'duration_seconds', 'resolution', 'file_size_mb']].head(10))"
        ),
        new_code_cell(
            "# Create visualizations\n"
            "fig, axes = plt.subplots(2, 2, figsize=(12, 10))\n"
            "fig.suptitle('Video Dataset Analysis')\n"
            "\n"
            "# Duration distribution\n"
            "df.boxplot(column='duration_seconds', by='label', ax=axes[0,0])\n"
            "axes[0,0].set_title('Duration by Label')\n"
            "axes[0,0].set_xlabel('Label')\n"
            "axes[0,0].set_ylabel('Duration (seconds)')\n"
            "\n"
            "# File size distribution\n"
            "df.boxplot(column='file_size_mb', by='label', ax=axes[0,1])\n"
            "axes[0,1].set_title('File Size by Label')\n"
            "axes[0,1].set_xlabel('Label')\n"
            "axes[0,1].set_ylabel('File Size (MB)')\n"
            "\n"
            "# Resolution distribution\n"
            "resolution_counts = df['resolution'].value_counts()\n"
            "axes[1,0].bar(range(len(resolution_counts)), resolution_counts.values)\n"
            "axes[1,0].set_title('Resolution Distribution')\n"
            "axes[1,0].set_xticks(range(len(resolution_counts)))\n"
            "axes[1,0].set_xticklabels(resolution_counts.index, rotation=45)\n"
            "\n"
            "# Frame count distribution\n"
            "df.hist(column='frame_count', bins=10, ax=axes[1,1])\n"
            "axes[1,1].set_title('Frame Count Distribution')\n"
            "axes[1,1].set_xlabel('Frame Count')\n"
            "axes[1,1].set_ylabel('Frequency')\n"
            "\n"
            "plt.tight_layout()\n"
            "plt.show()\n"
            "\n"
            "print('\\nDataset Statistics:')\n"
            "print(df.groupby('label')[['duration_seconds', 'file_size_mb', 'frame_count']].describe())"
        ),
        new_code_cell(
            "# Display sample frames from videos\n"
            "fig, axes = plt.subplots(2, 3, figsize=(15, 10))\n"
            "fig.suptitle('Sample Frames from Videos')\n"
            "\n"
            "# Show frames from real videos\n"
            "if real_videos:\n"
            "    frames = processor.extract_frames(real_videos[0], max_frames=3)\n"
            "    for i, frame in enumerate(frames):\n"
            "        axes[0, i].imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))\n"
            "        axes[0, i].set_title(f'Real Video - Frame {i+1}')\n"
            "        axes[0, i].axis('off')\n"
            "\n"
            "# Show frames from fake videos\n"
            "if fake_videos:\n"
            "    frames = processor.extract_frames(fake_videos[0], max_frames=3)\n"
            "    for i, frame in enumerate(frames):\n"
            "        axes[1, i].imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))\n"
            "        axes[1, i].set_title(f'Fake Video - Frame {i+1}')\n"
            "        axes[1, i].axis('off')\n"
            "\n"
            "plt.tight_layout()\n"
            "plt.show()\n"
            "\n"
            "print('✅ Video analysis complete!')"
        ),
    ],
    metadata={
        "kernelspec": {
            "display_name": "Python 3",
            "language": "python",
            "name": "python3"
        }
    }
)

with open("notebooks/02_video_analysis.ipynb", "w") as f:
    nbformat.write(nb, f)
print("✅ Notebook created at notebooks/02_video_analysis.ipynb")
```
